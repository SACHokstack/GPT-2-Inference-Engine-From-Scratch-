```python
import torch
import torch.nn as nn 

class MultiHeadAttention(nn.Module):

    def __init__(self,hidden,heads,max_seq_len):
        super().__init__()

        self.hidden = hidden
        self.heads = heads
        self.head_dim =hidden //heads

        self.register_buffer("causal_mask",torch.tril(torch.ones(max_seq_len,max_seq_len)))

        self.qkv_projection = nn.Linear(hidden,3 * hidden)
        self.output_projection =nn.Linear(hidden,hidden)



    def forward(self,x):
        B,T,C = x.shape


        qkv =self.qkv_projection(x)

        Q,K,V = torch.chunk(qkv,chunks=3,dim =-1)

        Q =Q.reshape(B,T,self.heads,self.head_dim)
        K =K.reshape(B,T,self.heads,self.head_dim)
        V =V.reshape(B,T,self.heads,self.head_dim)

        Q = Q.transpose(2,1)
        K =K.transpose(2,1)
        V =V.transpose(2,1)

        mask_slice = self.causal_mask[:T, :T]

        attention_output, _ = causal_attention(Q,K,V,mask_slice)

        attention_output =attention_output.transpose(2,1)

        attention_output =attention_output.reshape(B,T,C)

        output =self.output_projection(attention_output)

        return output



```

```python
class MyLayerNorm(nn.Module):

    def __init__(self,hidden_size,eps =1e-5):

        super().__init__()

        self.eps =eps

        self.weight = nn.Parameter(torch.ones(hidden_size))

        self.bias =nn.Parameter(torch.zeros(hidden_size))

    def forward(self, x):

        mean = torch.mean(x, dim=-1, keepdim=True)

        variance = torch.mean((x - mean) ** 2, dim=-1, keepdim=True)

        x_norm = (x - mean) / torch.sqrt(variance + self.eps)

        return x_norm * self.weight + self.bias

```

```python
torch.manual_seed(42)

x = torch.randn(2, 8, 768)

my_ln = MyLayerNorm(768)
torch_ln = nn.LayerNorm(768)

# Make both have exactly the same parameters
torch_ln.weight.data.copy_(my_ln.weight.data)
torch_ln.bias.data.copy_(my_ln.bias.data)

y_my = my_ln(x)
y_torch = torch_ln(x)

print("My shape:", y_my.shape)
print("PyTorch shape:", y_torch.shape)

print("Max difference:", torch.max(torch.abs(y_my - y_torch)))
print("Match:", torch.allclose(y_my, y_torch))
```

```python
class MLP(nn.Module):
    def __init__(self, hidden_size):
        
        super().__init__()

        self.fc1 = nn.Linear(hidden_size, 4 * hidden_size)
        self.gelu = nn.GELU(approximate="tanh")
        self.fc2 = nn.Linear(4 * hidden_size, hidden_size)

    def forward(self, x):
        x = self.fc1(x)
        x = self.gelu(x)
        x = self.fc2(x)
        return x
```

```python
import torch
import torch.nn as nn

class GPT2Block(nn.Module):

    def __init__(self, hidden_size, num_heads, max_seq_len):
        super().__init__()

        self.ln1 = MyLayerNorm(hidden_size)

        self.attention = MultiHeadAttention(hidden_size, num_heads, max_seq_len)
  
        self.ln2 = MyLayerNorm(hidden_size)
  
        self.mlp = MLP(hidden_size)

    def forward(self, x):
        
        residual = x
        
        x = self.ln1(x)
        x = self.attention(x)
        
        x = residual + x

        residual = x
    
        x = self.ln2(x)
        x = self.mlp(x)
        x = residual + x

        return x
```

```python
class GPT2Embeddings(nn.Module):

    def __init__(self, vocab_size, hidden_size, max_seq_len):
        super().__init__()

        self.token_embedding = nn.Embedding(vocab_size,hidden_size)

        self.position_embedding = nn.Embedding(max_seq_len,hidden_size)

    def forward(self, input_ids):

        # input_ids [B,T] 

        token_emb = self.token_embedding(input_ids)

        # Create positions 0,1,2,...,T-1

        positions = torch.arange(0,input_ids.shape[-1],device=input_ids.device)
        

        position_emb = self.position_embedding(positions)

        return token_emb + position_emb
```

```python
class GPT2(nn.Module):

    def __init__(self, vocab_size, hidden_size,
                 num_layers, num_heads, max_seq_len):

        super().__init__()

        self.embeddings = GPT2Embeddings(vocab_size,hidden_size,max_seq_len)

        self.blocks = nn.ModuleList([
            GPT2Block(hidden_size, num_heads, max_seq_len) 
            for _ in range(num_layers)
        ])

        self.ln_f = MyLayerNorm(hidden_size)

        self.lm_head = nn.Linear(hidden_size,vocab_size,bias=False)

    def forward(self, input_ids):

        x = self.embeddings(input_ids)


        for block in self.blocks:
            x = block(x)

        x = self.ln_f(x)

        logits = x @ self.embeddings.token_embedding.weight.T

        return logits
```

```python
@torch.no_grad()
def load_gpt2_weights(custom_model, hf_model):

    # embeddings
    custom_model.embeddings.token_embedding.weight.copy_(
        hf_model.transformer.wte.weight
    )

    custom_model.embeddings.position_embedding.weight.copy_(
        hf_model.transformer.wpe.weight
    )

    # transformer blocks
    for custom_block, hf_block in zip(
        custom_model.blocks,
        hf_model.transformer.h
    ):
        # LayerNorm 1
        custom_block.ln1.weight.copy_(hf_block.ln_1.weight)
        custom_block.ln1.bias.copy_(hf_block.ln_1.bias)

        # Attention QKV
        custom_block.attention.qkv_projection.weight.copy_(
            hf_block.attn.c_attn.weight.T
        )
        custom_block.attention.qkv_projection.bias.copy_(
            hf_block.attn.c_attn.bias
        )

        # Attention output projection
        custom_block.attention.output_projection.weight.copy_(
            hf_block.attn.c_proj.weight.T
        )
        custom_block.attention.output_projection.bias.copy_(
            hf_block.attn.c_proj.bias
        )

        # LayerNorm 2
        custom_block.ln2.weight.copy_(hf_block.ln_2.weight)
        custom_block.ln2.bias.copy_(hf_block.ln_2.bias)

        # MLP
        custom_block.mlp.fc1.weight.copy_(
            hf_block.mlp.c_fc.weight.T
        )
        custom_block.mlp.fc1.bias.copy_(
            hf_block.mlp.c_fc.bias
        )

        custom_block.mlp.fc2.weight.copy_(
            hf_block.mlp.c_proj.weight.T
        )
        custom_block.mlp.fc2.bias.copy_(
            hf_block.mlp.c_proj.bias
        )

    # final LayerNorm
    custom_model.ln_f.weight.copy_(
        hf_model.transformer.ln_f.weight
    )
    custom_model.ln_f.bias.copy_(
        hf_model.transformer.ln_f.bias
    )
```

```python
custom_model = GPT2(
    vocab_size=50257,
    hidden_size=768,
    num_layers=12,
    num_heads=12,
    max_seq_len=1024,
)

custom_model.eval()

load_gpt2_weights(custom_model, hf_model)
```

```python
hf_outputs = {}
custom_outputs = {}

def save_output(storage, name):
    def hook(module, inputs, output):
        hf_or_custom_output = output[0] if isinstance(output, tuple) else output
        storage[name] = hf_or_custom_output.detach()
    return hook

handles = []

for i in range(12):
    handles.append(
        hf_model.transformer.h[i].register_forward_hook(
            save_output(hf_outputs, f"block_{i}")
        )
    )

    handles.append(
        custom_model.blocks[i].register_forward_hook(
            save_output(custom_outputs, f"block_{i}")
        )
    )

with torch.no_grad():
    hf_logits = hf_model(input_ids).logits
    custom_logits = custom_model(input_ids)

for handle in handles:
    handle.remove()

for i in range(12):
    diff = (
        hf_outputs[f"block_{i}"]
        - custom_outputs[f"block_{i}"]
    ).abs().max().item()

    print(f"Block {i:2d}: max diff = {diff:.8f}")

print(
    "Final logits:",
    (hf_logits - custom_logits).abs().max().item()
)
```
